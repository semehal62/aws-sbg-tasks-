# Week 1 Lab — Student Grade Data Processing

**Goal:** Load a raw CSV, clean it with Python functions, and analyze it with NumPy.

This notebook uses Week 1 concepts: variables, lists, dictionaries, loops, conditions, and functions. NumPy is used for numerical calculations.


```python
from data_loader import load_data, clean_data
from analyzer import subject_averages, student_averages, pass_rate, top_student
```

## 1. Load the raw dataset

```python
raw_students = load_data("data/students.csv")

print("Number of raw records:", len(raw_students))
print("First raw record:")
print(raw_students[0])
```

The CSV intentionally contains blank grades, invalid values, and a duplicate student record so we can practice cleaning.

## 2. Clean the data

```python
cleaned_students = clean_data(raw_students)

print("Records after removing duplicates:", len(cleaned_students))
print("First cleaned record:")
print(cleaned_students[0])
```

## 3. Analyze the cleaned data with NumPy

The analyzer converts the grade values into a two-dimensional NumPy array. Each row represents a student, and each column represents a subject.

```python
import numpy as np
from analyzer import get_score_array

score_array = get_score_array(cleaned_students)

print("NumPy score array:")
print(score_array)
print("Array shape (students, subjects):", score_array.shape)
```

## 4. Calculate subject averages

```python
averages = subject_averages(cleaned_students)

for subject in averages:
    print(subject.title(), "average:", round(averages[subject], 2))
```

## 5. Calculate student averages and find the top student

```python
averages_by_student = student_averages(cleaned_students)

for student in averages_by_student:
    print(student["name"], "average:", round(student["average"], 2))

best = top_student(cleaned_students)
print("\nTop student:", best["name"])
print("Top average:", round(best["average"], 2))
```

## 6. Calculate the pass rate

```python
rate = pass_rate(cleaned_students, passing_score=50)
print("Class pass rate:", round(rate, 2), "%")
```

## 7. What did we learn?

- CSV files can be loaded into lists of dictionaries.
- Functions make cleaning steps reusable.
- Loops and conditions help identify invalid values and duplicates.
- NumPy arrays make numerical calculations easier.
- `axis=0` calculates averages down each subject column; `axis=1` calculates averages across each student's subjects.

**Data-cleaning note:** filling missing grades with the subject average is a simple demonstration. It is not always the right choice for real datasets.
